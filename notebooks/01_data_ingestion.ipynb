Load all 10 CSV datasets using pandas, print shape/dtypes/head for each

```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd() / "mutual_fund_project"
raw_path = project_root / "data" / "raw"

csv_files = sorted(raw_path.glob("*.csv"))

datasets = {}

for file in csv_files:
    print("----------")
    print(f"FILE: {file.name}")
    print("----------")

    df = pd.read_csv(file)

    datasets[file.stem] = df

    print("\nShape:")
    print(df.shape)

    print("\nData Types:")
    print(df.dtypes)

    print("\nFirst 5 Rows:")
    print(df.head())

    print("\n\n")
```
